## Stage 2 – Data Cleaning and Pre-processing

- Handle missing values (impute or drop)

- Handle duplicates
- column name rename or dropping unwanted columns if required

- Treat outliers if required

- Check skewness and apply transformations

- Convert data types if needed

- Feature transformations (date parts, derived fields if required for analysis)

```python
# Import Pandas and load the dataset
import pandas as pd
import numpy as np
df=pd.read_csv("https://raw.githubusercontent.com/020521/E-commerce-analysis/refs/heads/main/ecommerce_customer_behavior_dataset.csv")
df.head()
```

```python
# check your column names
print(df.columns)
```

```python
# Convert all column names to lowercase

df.columns = df.columns.str.lower()
```

```python
# rename cloumn name
df.rename(columns={
    'date': 'order_date',
    'total_amount': 'sales_amount'
}, inplace=True)

print(df.columns)
```

```python
# Check duplicate order IDs separately:

duplicate_orders = df[df['order_id'].duplicated(keep=False)]
print("Total duplicate Order IDs:", df['order_id'].duplicated().sum())
print(duplicate_orders)
```

```python
# Remove extra spaces from text columns

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

```python
# Check existing data types:

df.dtypes
```

```python
# check outliers
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, ":", len(outliers), "possible outliers")
```

```python
# Check skewness
df.select_dtypes(include='number').skew()
```

```python
# Apply Log Transformation to sales_amount

df['log_sales_amount'] = np.log1p(df['sales_amount'])

print("Before Transformation:")
print(df['sales_amount'].skew())

print("After Transformation:")
print(df['log_sales_amount'].skew())
```

```python
# Change the date column datatype

df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'].dtype
df['date'].isnull().sum()
```

```python
# Extract Year,moth,date
print(df.columns.tolist())
df.columns = df.columns.str.strip().str.lower()
print(df.columns.tolist())

df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()

print(df[['order_date', 'year', 'month', 'month_name']].head(5))

print(df['month_name'].unique())
```

```python
# Check date range
print("Minimum Date:", df['date'].min())
print("Maximum Date:", df['date'].max())
```

```python
# check invalid ages
print(df[df['age'] <= 0])
```

```python
# Check how many orders are in each month

print(df['month_name'].value_counts())
```

```python
# Check negative prices and quantities
print(df[df['unit_price'] < 0])

print(df[df['quantity'] <= 0])
```

```python
# Check customer ratings
# If the rating scale is 1 to 5

print(df['customer_rating'].unique())
print(df[(df['customer_rating'] < 1) |
         (df['customer_rating'] > 5)])
```

```python
# 1. Univariate Analysis
# 1.1 Countplot – Gender Distribution

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('https://raw.githubusercontent.com/020521/E-commerce-analysis/refs/heads/main/ecommerce_customer_behavior_dataset.csv')

print(df.head())
print(df.shape)

plt.figure(figsize=(7,5))

sns.countplot(data=df, x='Gender')

plt.title('Customer Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Orders')
plt.show()
```

Interpretation:
- Identify which gender has the highest number of orders.
- Compare the number of male and female customers.
- This helps understand customer demographics.

Business insight: The business can use customer demographics to plan targeted marketing campaigns.

```python
# 1.2 Histogram – Age Distribution
# Which age group places more orders?

plt.figure(figsize=(8,5))

sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

Interpretation:
- Identify the age range with the highest number of customers.
- Check whether customers are concentrated in younger or older age groups.
- Observe the overall age distribution.

Business insight: Understanding customer age groups helps the company target products and promotions to the right audience.

```python
# 1.3 Boxplot – Sales Amount
# Are there unusually high or low sales amounts?

plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='sales_amount')

plt.title('Sales Amount Distribution')
plt.xlabel('Sales Amount')
plt.show()
```

Interpretation:
- The middle line represents the median sales amount.
- The box represents the middle 50% of values.
- Points outside the whiskers may indicate outliers.
- Check whether sales amounts are spread widely.

Business insight: Identifying unusually high or low transactions helps the business understand spending patterns and investigate exceptional orders.

```python
# 2. Bivariate Analysis
# 2.1 Scatterplot – Quantity vs Sales Amount
# Does purchasing more products increase sales?

plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='quantity', y='sales_amount')

plt.title('Quantity vs Sales Amount')
plt.xlabel('Quantity')
plt.ylabel('Sales Amount')
plt.show()
```

Interpretation:
- Observe whether sales increase when quantity increases.
- Check whether the points follow an upward or downward pattern.
- Identify unusual transactions.

Business insight: This helps understand how the number of products purchased affects revenue.

```python
# 2.2 Barplot – Revenue by Product Category
# Which product category generates the highest revenue?

plt.figure(figsize=(10,5))

sns.barplot(data=df, x='product_category',
            y='sales_amount', estimator='sum')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

Interpretation:
- Identify the category generating the highest revenue.
- Compare revenue across different categories.
- Identify categories contributing less to total revenue.

Business insight: The company can focus on high-revenue categories and improve promotions for low-performing categories.


```python
# 2.3 Correlation Heatmap – Numerical Columns
# Which numerical variables have strong relationships?

plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include='number')

sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

Interpretation:
- Values close to +1 indicate strong positive correlation.
- Values close to -1 indicate strong negative correlation.
- Values close to 0 indicate weak linear correlation.

Business insight: Correlation helps identify which factors move together, such as quantity and sales amount. However, correlation does not prove causation.


```python
# 3. Multivariate Analysis
# 3.1 Pairplot – Numerical Variables
# How are age, quantity, unit price and sales amount related?

selected_columns = [
    'age',
    'quantity',
    'unit_price',
    'sales_amount'
]

sns.pairplot(df[selected_columns].sample(
    min(500, len(df)), random_state=42
))

plt.show()
```

Interpretation:
- Compare relationships between multiple numerical columns.
- Identify positive or negative patterns.
- Observe the distribution of individual variables along the diagonal.

Business insight: This helps the company understand how customer age, product price and purchase quantity relate to sales.

```python
# 3.2 Grouped Analysis – Revenue by Gender and Category
# Which product categories generate more revenue from male and female customers?

gender_category = df.groupby(
    ['gender', 'product_category']
)['sales_amount'].sum().reset_index()

plt.figure(figsize=(12,6))

sns.barplot(
    data=gender_category,
    x='product_category',
    y='sales_amount',
    hue='gender'
)

plt.title('Revenue by Gender and Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

Interpretation:
- Compare category revenue between genders.
- Identify which gender contributes more revenue to each category.
- Identify differences in purchasing patterns.

Business insight: The company can develop targeted promotions based on customer preferences and category performance.

```python
# 3.3 Pivot Table – Monthly Revenue Analysis
# How does monthly revenue change across different years?

monthly_revenue = pd.pivot_table(
    df,
    values='sales_amount',
    index='year',
    columns='month_name',
    aggfunc='sum',
    fill_value=0
)

display(monthly_revenue)
```

Interpretation:
- Compare monthly revenue across different years.
- Identify months with the highest and lowest revenue.
- Observe seasonal sales patterns.

Business insight: The business can plan inventory, discounts and marketing campaigns based on monthly sales trends.

```python
# 3.4 Advanced Plot – Returning Customer Analysis
# Do returning customers generate more revenue than new customers?

plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='is_returning_customer',
    y='sales_amount',
    estimator='mean'
)

plt.title('Average Sales by Customer Type')
plt.xlabel('Returning Customer')
plt.ylabel('Average Sales Amount')
plt.show()
```

Interpretation:
- Compare the average sales amount of returning and new customers.
- Identify which customer group spends more per order.

Business insight: If returning customers have higher average order values, loyalty programmes and retention campaigns may help increase revenue.

```python
# 1. KPI Metrics – Sales Performance
total_revenue = df['sales_amount'].sum()

total_orders = df['order_id'].nunique()

total_customers = df['customer_id'].nunique()

total_units_sold = df['quantity'].sum()

average_order_value = total_revenue / total_orders

total_cities = df['city'].nunique()

print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Units Sold:", total_units_sold)
print("Average Order Value:", round(average_order_value, 2))
print("Total Cities:", total_cities)
```

```python
# Sales by City Chart

city_sales = df.groupby('city')['sales_amount'].sum().reset_index()

city_sales = city_sales.sort_values(
    by='sales_amount', ascending=False
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=city_sales,
    x='city',
    y='sales_amount'
)

plt.title('Total Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
# Display Metrics as KPI Cards
fig, axes = plt.subplots(2, 3, figsize=(14, 7))

metrics = [
    ('Total Revenue', df['sales_amount'].sum()),
    ('Total Orders', df['order_id'].nunique()),
    ('Total Customers', df['customer_id'].nunique()),
    ('Units Sold', df['quantity'].sum()),
    ('Average Order Value', df['sales_amount'].sum() / df['order_id'].nunique()),
    ('Total Cities', df['city'].nunique())
]

for ax, (title, value) in zip(axes.flatten(), metrics):
    ax.text(
        0.5, 0.6, f'{value:,.2f}',
        ha='center',
        fontsize=18,
        fontweight='bold'
    )
    ax.set_title(title, fontsize=12)
    ax.axis('off')

plt.suptitle('E-commerce Business Performance Dashboard',
             fontsize=16)

plt.tight_layout()
plt.show()
```

Interpretation
- Total Revenue: Shows the total sales generated by the business.
- Total Orders: Shows the number of unique orders placed.
- Total Customers: Shows the number of unique customers.
- Units Sold: Shows the total quantity of products purchased.
- Average Order Value: Shows the average amount spent per order.
- Total Cities: Shows the number of cities covered by the business.

Business Insights
1. Understand the overall sales performance.
2. Measure customer purchasing activity.
3. Identify the size of the customer base.
4. Understand product demand.
5. Evaluate average customer spending.
6. Measure geographical business coverage.

```python
# Descriptive Analysis
print("Total Revenue:", df['sales_amount'].sum())

print("Total Orders:", df['order_id'].nunique())

print("Total Customers:", df['customer_id'].nunique())

print("Average Order Value:",
      df['sales_amount'].sum() / df['order_id'].nunique())

print("Average Customer Rating:",
      df['customer_rating'].mean())

```

```python
# Revenue by Product Category
category_revenue = df.groupby('product_category')['sales_amount'].sum()

display(category_revenue.sort_values(ascending=False))
```

```python
#Orders by City

city_orders = df.groupby('city')['order_id'].nunique()

display(city_orders.sort_values(ascending=False))
```

Interpretation: Descriptive analysis helps understand overall revenue, orders, customers and product performance.

```python
# 2. Diagnostic Analysis
# Investigate the reasons behind business performance.

# Revenue by Payment Method
payment_analysis = df.groupby('payment_method').agg(
    total_revenue=('sales_amount', 'sum'),
    total_orders=('order_id', 'nunique')
).reset_index()

display(payment_analysis)
```

```python
# Customer Rating by Product Category

rating_analysis = df.groupby('product_category').agg(
    average_rating=('customer_rating', 'mean'),
    total_orders=('order_id', 'count')
).reset_index()

display(rating_analysis)
```

```python
# Delivery Time vs Customer Rating

delivery_analysis = df.groupby('delivery_time_days').agg(
    average_rating=('customer_rating', 'mean'),
    total_orders=('order_id', 'count')
).reset_index()

display(delivery_analysis)
```

Interpretation: Diagnostic analysis helps investigate how payment methods, product categories and delivery times are associated with business performance.

```python
# 3. Predictive Analysis
# Monthly Revenue

monthly_revenue = df.groupby('month')['sales_amount'].sum().reset_index()

display(monthly_revenue)
```

```python
# Simple Revenue Forecast
from sklearn.linear_model import LinearRegression

X = monthly_revenue[['month']]
y = monthly_revenue['sales_amount']

model = LinearRegression()
model.fit(X, y)

future_months = pd.DataFrame({'month': [13, 14, 15]})

predicted_revenue = model.predict(future_months)

forecast = future_months.copy()
forecast['predicted_revenue'] = predicted_revenue

display(forecast)
```

```python
# 4. Prescriptive Analysis
# This section converts your findings into business recommendations.
# Identify High and Low Revenue Categories
category_revenue = df.groupby('product_category')['sales_amount'].sum()

highest_category = category_revenue.idxmax()
lowest_category = category_revenue.idxmin()

print("Highest Revenue Category:", highest_category)
print("Lowest Revenue Category:", lowest_category)
```

```python
# Business Recommendations
print("BUSINESS RECOMMENDATIONS")

print("1. Focus marketing on the highest revenue-generating category.")

print("2. Develop promotional offers for lower-performing categories.")

print("3. Improve delivery operations to support customer satisfaction.")

print("4. Monitor customer ratings to identify service improvements.")
```
